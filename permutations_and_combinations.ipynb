## Arrangements
The number of ways to arrange n objects: n!

In a circle: (n - 1)!

$$ \frac{n!}{j!k!} $$
Arrange k objects where j objects and k objects are alike.

n! number of arrangements we treat n objects as distinct, which includes j! number of repeated counts e.g. h1h2 and h2h1 are the same and should only count 1.

## Permutations
The total number of ways we can order r objects chosen from a larger
set of n objects.
$$ {}^{n}P_{r} = \frac{n!}{(n-r)!} $$

2 ways of thinking:
1. Order r objects: the first has n choices, the second has (n-1)
   choices, ..., the r-th has (n-r+1) choices. So the result is
   $n \times (n-1) \times \cdots \times (n-r+1)$.
2. Start from all n! orderings of n objects. The last (n-r) objects'
   internal order doesn't matter (they're not part of our r-object
   selection), and each distinct r-object arrangement is repeated
   (n-r)! times because of that. So divide n! by (n-r)! to remove
   the duplicates. e.g. n = 20, r = 3: each 3-object arrangement is
   paired with 17! orderings of the remaining 17 objects, which we
   don't care about, so we divide by 17!.

## Combinations
Take r objects from a large pool of n objects where order doesn't count.
$$ ^{n}C_{r} = \frac{n!}{r!(n-r)!} $$
Here r-pattern will repeat r! times so divided by it.
