# Definition
Statistics are numbers that summarize raw facts and figures in some
meaningful way.

# Categories vs Quantities
## Categorial or qualitative
Data is split into categories that describe qualities or 
characteristics. Data values can not be interpreted as numbers.
## Numerical or quantitative data
It’s data where the values have meaning as numbers, and 
that involves measurements or counts.

# Charts
## Pie chart 
good for display proportions

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.style.use('_mpl-gallery-nogrid')
x = [1,2,3,4]
colors = plt.get_cmap("Blues")(np.linspace(0.2, 0.7, len(x)))

fig, ax = plt.subplots()
ax.pie(x, colors=colors, radius=3, center=(4, 4),
       wedgeprops={"linewidth": 1, "edgecolor": "white"}, frame=True)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))
plt.show()
```

## Bar chart
* Horizontal bar charts are used for categorical data, 
particularly where the category names are long.
* Vertical bar charts are used for numerical data, or 
categorical data if the category names are short.
* Show frequencies and proportions using segmented bar chart.

## Histogram chart 
* For grouped numeric data
* No gaps between bars(1. to show that there are no gaps in the values, and that every value is covered. 2. so that the width of the interval reflects the range of the values you’re covering.)
* Frequency is equal to bar area
* For bars with different widths, frequency density(frequency / width) is a way to make the frequency proportional to the area of a bar, rather than height
* For discrete data we can expand the range of original data, e.g. 1 ranging from 0.5 to 1.5 and position 1 on the center
