## Expectation value
$$ E(X) = \Sigma x * P(X=x) $$

It's the mean for unhappened events. For example, toss a dice get 1,2,3...6, mean will be (1+2+3...6)/6, E(X) = $$ \frac{1}{6} * 1 + \frac{1}{6} * 2 + ... \frac{1}{6} * 6 $$

\begin{align*}
Var(X) &= E(X - \mu)^2, \quad \mu = E(X) \\
&= \sum (x- \mu)^{2} P(X=x) \\
&= \sum (x^{2} - 2 x \mu + \mu^{2} )P(X=x)\\
&= \sum x^{2}P(X=x) -2 \sum x \mu P(X=x) + \sum \mu^{2}P(X=x)\\
&= E(X^{2}) - 2 \mu^{2} + \mu^{2} \\
&= E(X^{2}) - E(X)^{2}\\
\end{align*}

## Independent variables
$$
\begin{align*}
E(X+Y) = E(X) + E(Y)\\
E(X-Y) = E(X) - E(Y)\\
Var(X+Y) = Var(X) + Var(Y)\\
Var(X-Y) = Var(X) + Var(Y)\\
\end{align*}
