## Expectation value
$$ E(X) = \Sigma x * P(X=x) $$

It's the mean for unhappened events. What E(X) represents depend on the random variable X. For example, toss a dice get 1,2,3...6, mean will be (1+2+3...6)/6, E(X) = $$ \frac{1}{6} * 1 + \frac{1}{6} * 2 + ... \frac{1}{6} * 6 $$

\begin{align*}
Var(X) &= E(X - \mu)^2, \quad \mu = E(X) \\
&= \sum (x- \mu)^{2} P(X=x) \\
&= \sum (x^{2} - 2 x \mu + \mu^{2} )P(X=x)\\
&= \sum x^{2}P(X=x) -2 \sum x \mu P(X=x) + \sum \mu^{2}P(X=x)\\
&= E(X^{2}) - 2 \mu^{2} + \mu^{2} \\
&= E(X^{2}) - E(X)^{2}\\
\end{align*}

## Linear transofrmations

\begin{align*}
E(aX + b) &= \frac{\Sigma (ax+b) P(X=ax+b)}{n}\\
&=\frac{\Sigma ax P(X=ax+b)+b P(X=ax+b)}{n}\\
&=\frac{a\Sigma x P(X=ax+b) + b}{n}\\
&=aE(X)+b\\
\end{align*}

## Independent variables
$$
\begin{align*}
E(X+Y) = E(X) + E(Y)\\
\text {The number of Xs must be the same as that of Y so that X + Y can defined as a random variable.} \\
\text {e.g. Roll 2 dices, if we roll first dice 3 times, and roll second one 4 times, then thre's no match for the fourth throwing for second dice.}\\
E(X-Y) = E(X) - E(Y)\\
Var(X+Y) &= \frac{(X+Y-\mu_{1} -\mu_{2})^{2}}{n}\\
&=\frac{(X-\mu_{1})^{2} + (Y-{\mu_2})^{2} - 2 (X-\mu_{1})(Y-{\mu_2})}{n}\\
&=Var(X) + Var(Y)\\
Var(X-Y) = Var(X) + Var(Y)\\
\end{align*}
