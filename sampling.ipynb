## Point estimator
A **best guess** for the value of a population parameter, derived from sample data. Add **^** symbol to represent it e.g. $ \hat{\mu} $, $\hat{\sigma^{2}}$

$$ \hat{\mu} = \bar{x} $$
$$ \hat{\sigma^{2} = S^{2} = \frac{\Sigma (X - \bar{X}^{2})}{n-1}} $$

## Population proportion
p is the proportion of success within the population. 
$$\hat{p} = p_{s}$$
$$p_{s} = \frac{\text{number of successes}}{\text{number in sample}}$$

## Having population statistics, calculating sample statistics
Core: Derive the mean, variance of sampling distribution to calculate prob.

### Having p, calculate the probability for $p_s$
Q: Knowing the proportion of red candies is 20% in population, what's the prob for a packet having more than 40% red candies.

**Sampling distribution of proportions**: forming a distribution out
of the sample proportion $p_s = X/n$ across all possible samples.

Because $X \sim B(n,p)$, we get $E(p_s) = p$, $Var(p_s) = \frac{pq}{n}$.
Then when n is large (CLT), $p_s \sim N\left(p, \frac{pq}{n}\right)$
approximately, and we calculate probabilities by standardizing and
looking up the normal probability table (apply continuity correction,
since $X$ is discrete but we're approximating with a continuous
distribution).

### Having population $\mu$ and $\sigma$, calculate probabilities for sample mean

$$ \bar{x} = \frac{x_{1} + x_{2} + ... + x_{n}}{n} $$

$$
\begin{align*}
E(\bar{x}) &= E\left(\frac{x_{1} + x_{2} + ... + x_{n}}{n}\right)\\
&=\frac{1}{n}(E(x_{1})+E(x_{2})+...+E(x_{n}))\\
&=\frac{1}{n} \times n \times \mu \\
&= \mu
\end{align*}
$$

$$
Var(\bar{x}) = \frac{\sigma^{2}}{n}
$$

If $X \sim N(\mu, \sigma^{2})$ then $\bar{x} \sim N(\mu, \frac{\sigma^2}{n})$
Otherwise by CLT: if n is large (e.g. n ≥ 30), $\bar{x}$ approximately follows a
normal distribution **regardless of the population's original
distribution shape**: $\bar{x} \sim N\left(\mu, \frac{\sigma^2}{n}\right)$.
