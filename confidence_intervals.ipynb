## Why need it
Even though **point estimators** give a "best guess" for a population
parameter derived from an unbiased sample, they can't tell us how
close that guess is to the true value. A **confidence interval**
gives a way to express how confident we are that the true value
falls within a range.

## Confidence interval
A range of values, calculated from sample data, that is likely to
contain the true **population parameter** (e.g. population mean μ or
proportion p).

## Confidence level
The proportion of confidence intervals, constructed this way over
repeated sampling, that would contain the true population parameter
(e.g. 95% confidence level → if we repeated this sampling process many
times, about 95% of the resulting intervals would contain the true
parameter).

## T-distibution
If $X \sim N(\mu, \sigma^{2})$ and n is small, then $\bar{X}\sim t(\nu)\quad \nu\text{ is the degree of freedoms}$

## General form of confidence interval
$$ statistic \pm \text{margin of error} $$
