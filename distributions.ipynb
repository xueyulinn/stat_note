## Overview
Special distributions for definite patterns, once knowing the distribution we can use a few shortcuts: much easier to calculate probabilities, expectations and variances.

## Geometric distribution
Covers situations where:
* Trials are independent
* Trial is either success or failure
* The thing we're interested is how many trials are needed for success
$$ P(X=r) = p \times q^{r-1} $$
$$ P(X <= r) + P(X > r) = 1 $$
$$ P(X <= r) = 1 - q^{r} $$

If variable X follows a geometric distribution, it can be written as $ X \sim Geo(p) $
$$ E(X) = \frac{1}{p} \\ Var(X) = \frac{q}{p^{2}}\\ $$
