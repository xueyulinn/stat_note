## Overview
Probability distribution lists **all possible values of a random variable X and corresponding probabilities**. There are special distributions for definite patterns, once knowing the distribution we can use a few shortcuts: much easier to calculate probabilities, expectations and variances.

## Geometric distribution
Covers situations where:
* Trials are independent
* Trial is either success or failure
* The thing we're interested is how many trials are needed for success

### Probability
$$ P(X=r) = p \times q^{r-1} $$
$$ P(X <= r) + P(X > r) = 1 $$
$$ P(X <= r) = 1 - q^{r} $$

### Expectation and Variance
If variable X follows a geometric distribution, it can be written as $ X \sim Geo(p) $
$$ E(X) = \frac{1}{p} \\ Var(X) = \frac{q}{p^{2}}\\ $$

## Binominal distribution
Covers situations where:
* Trials are independent
* Trial is either success or failure
* The number of trials is finite

$$ X \sim B(n,p) $$

### Probability
$$P(X=r) = ^{n}C_{r}P^{r}q^{n-r}\\$$

### Expectation
Expectation E(X) X here is the numer of success out of n.

Suppose we only do one trial, it will be either failure or success: $E(X) = 0 * q + 1 * p = p$, all trials are independent so $E(X) = np$.

### Variance

\begin{align*}
Var(X) &= E(X^{2}) - E(X)^{2}\\
&= p - p^{2}\\
&= p(1-p)\\
&= pq\\

Var(X) = npq\\
\end{align*}

## Poisson distribution
Cases:
* Individual events occur randomly at an interval.
* We know the mean number of occurences in the given interval $\lambda$.

X is the number of occurences in a given interval.
### Probability
$$ P(X=r) = \frac{e^{-\lambda} \lambda^{r}}{r!} $$

### Expectation and variance
$$ E(X) = \lambda $$
$$ Var(X) = \lambda $$

### Combine Poisson variables
$$ X + Y \sim(\lambda_{x} + \lambda_{y}) $$

### Approximate Binominal distribution with Possion distribution

$\lambda$ to be like np

$\lambda$ to be like npq

np to be like npq when q is nearly 1

X $\sim B(n,p)$ can be approximated by X $\sim$ Po(np) if n is large and p is small

## Normal distribution
All above distributions are about discrete data(取值可数) which are distinct numeric values. We have continuous data deals with **measurement**. What we're talking about is acceptable degrees of accuracy.

When dealing with continuous data, we calculate probabilities for a **range of values**. We can't get all values within a range like we did for discrete data, it has infinite values.

Using **probability density funcntion** to describe the probability of a random variable.

The probability for a single precise value is 0. 2 ways of thinking about this:
1. Width is zero -> area is zero.
2. The probability of getting a precise value to infite number of values. e.g. To get a precise 10-inch piece of string, we need a measurement of 10 inch long to the nearest atom and examined with a powerful microscope.

### Probability
1. If $X \sim N(\mu, \sigma^{2})$, gets the mean and variance and the probability range
2. standarlize original distribution to make $X \sim N(0, 1)$, using Z-score
3. look up probability table to resolve range probabilities

### Combined variables
X and Y are independent
$$ X + Y \sim(\mu, \sigma^{2})\\
\mu = \mu_{X} + \mu_{Y}\\
\sigma^{2} = \sigma_{X}^{2} + \sigma_{Y}^{2}
$$

### Linear transformation
$$ ax + b \sim N(a\mu+b, a^{2}\sigma^{2}) $$

### Approximation
Normal distribution can approximate Binomial and Poisson distributions
when their **shape becomes roughly symmetric and bell-shaped**:
- **Binomial**: approximation works well when n is large and p is not
  too close to 0 or 1 (common rule of thumb: np ≥ 5 and n(1−p) ≥ 5)
- **Poisson**: approximation works well when λ is large (e.g. λ ≥ 10),
  since the distribution becomes more symmetric as λ increases

**Geometric distribution** stays strongly right-skewed regardless of
parameter value, so it can't be approximated by a normal distribution.

#### Approximate binominal distribution
A general rule is when nq > 5 and nq > 5 $ X \sim B(n, p) \to X \sim N(np, npq) $.

We have to expand a discreate value into a range because a discreate value has no width so its area is alway 0, but in normal distribution we calculate probability with area. We apply rounding rule (round to the nearest integer). e.g. 3 is represented by [2.5, 3.5).

Apply **continuity correction** when approximating. 
* P(X <= a), using [a-0.5, a+0.5) to reprensent a, we need to calculate P(X < a+0.5) to include a.
* P(X < a), calculate P(X < a-0.5) not to include a. 
* P(X = 0), calculate P(-0.5 < x < 0.5)
