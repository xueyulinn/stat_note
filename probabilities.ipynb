## Event
An outcome that has a probability assigned to it.

## Probability
$$ P(A) = n(A) / n(S) $$
$$ P(A') + P(A) = 1 $$
A' is the complementary event of A

If two events mutually exclusive, only one of the two can occur.

If two events interact, it's possible they can occur simultaneouly.

### Notation
$$ A \cap B $$
refers the intersection between A and B. It takes the **common elements** of events.

$$ A \cup B $$
includes all of the elements in A and also those in B. If A U B = 1 then A and B are said to be **exhaustive**.
$$ P(A) \cup P(B) = P(A) + P(B) - P(A\cap B)  $$

A and Complementary A are both exclusive and exhaustive.

## Conditional probabilities
Measure the probability of one event occurring relative to another occuring.
$$ P(A|B) = P(A\cap B) / P(B) $$
$$ P(A\cap B) = P(A|B) * P(B) $$
$$ P(B|A) = P(B\cap A) / P(A) $$
$$ P(B\cap A) = P(B|A) * P(A) $$

## Bayes theorem
Calculate P(A|B) given P(A), P(B|A), P(B|A')
\begin{align*}
P(A|B) &= \frac{P(A \cap B)}{P(B)} \\
&= \frac{P(B \cap A)}{P(B)} \\
&= \frac{P(A) * P(B|A)}{P(B)} \\
&= \frac{P(A) * P(B|A)}{P(A \cap B) + P(A' \cap B)} \\
&= \frac{P(A) * P(B|A)}{P(B \cap A) + P(B \cap A')} \\
&= \frac{P(A) * P(B|A)}{P(A) * P(B|A) + P(A') * P(B|A')} \\
\end{align*}

## Dependent events && independent events
$$ P(A|B) != P(A) $$
A and B are dependent because the probability of A given B happens changes the probability of A.

\begin{align*}
 P(A|B) = P(A)\\
 P(A \cap B) = P(A) * P(B)\\ 
\end{align*}
A and B are independent.

Two mutually exclusive events are dependent because A occurring leads to B unoccurring.
