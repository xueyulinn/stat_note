# Central tendency

## Mean
$\mu = \Sigma x / n$

$\mu = \Sigma fx / \Sigma f$

Use when the data is fairly symmertric.

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
age = np.array([19,20,21,145,147])
freq = np.array([3,6,3,1,1])
plt.hist(age, weights=freq, bins=10)
plt.show()
```

The mean is 38 but nobody is around that age because of **outliers**.

```python
import seaborn as sns
data = np.repeat(age, freq)  # 先展开数据
sns.kdeplot(data)
plt.show()
```

Data right skewed

## Median
Middle value

Use when data is skewed because of outliers.
* If the data is skewed to the right mean is to the right of the median.
* If the data is skewed to the left mean is to the left of the median.

```python
age = np.array([1,2,3,31,32,33])
freq = np.array([3,4,2,2,4,3])
plt.bar(age,freq, width=1)
plt.show()
```

Both Mean and Median are 17 which are misleading.

# Spread

## Range
How data spread out?

Higher bound is the highest value and lower bound is the lowest value.

Range = higher bound - lower bound.

Sensitive to outliers.

## Quartiles
The values that split data into chunks, the lowest quartile is lower quartile, the highest quartile is upper quartile and the middle quartile is median.

Interquartile range = upper quartile - lower quartile

![quartiles](diagrams/quartiles.png)

Less sensitive to outliers as the interquartile range excludes the outliers.

计算:

n 个数据点,排序后一共占据 n 个"位置"。如果你想找到把数据切成 p 这个比例的点,理论上这个点应该在第 p×n 个位置附近。

Lower quartile(下四分位数,Q1)= 左边 25% → 位置 ≈ n × 1/4 = n/4

Upper quartile(上四分位数,Q3)= 左边 75% → 位置 ≈ n × 3/4 = 3n/4

If integer then quartile = (n/4 + n/4 + 1) / 2 else roundup(n/4)

## Percentiles
We can break up the set of data into percentages. Percentiles are the values that split data into percentages the same way split data into quartiles. Quartiles are a type of percentiles.

P(k) = k(n/100)

If P(k) is int then quartile = (P(k) + P(k) + 1) / 2 else roundup(P(k))

## Box and whisker diagrams/box plots
Box shows interquartile range and whiskers show lower and higher bounds.

## Variance
Interquartile range only uses part of the data for measuring spread while the range is only used for the difference between the highest and lowest bound. We need a way to figure out the distribution of all data.

Variance is the average of the distance of values from the mean squared.
$$ Variance = \Sigma(x - \mu)^2 / n $$

$$ Variance = (\Sigma(x^2) - \mu^2) / n $$

## Standard deviation
$$ \sigma = \sqrt[2]{variance} $$

## Z-score / standard score
A way to compare data points from different distributions (different means
and standard deviations) on the same scale.

$$z = \frac{x - \mu}{\sigma}$$

This transforms data into a standardized scale with mean 0 and standard
deviation 1 — **but the shape of the distribution is unchanged** (skewed
stays skewed, it does not become normal). The **relative position** of data points keeps unchanged. A data point is one std away from mean, it is still one std away from mean after standarlization. 

1. $x - \mu$ shifts the whole distribution so the mean becomes 0:
   $$\text{new } \mu = \frac{\sum(x-\mu)}{n} = \frac{\sum x}{n} - \mu = \mu - \mu = 0$$
2. Dividing by $\sigma$ rescales the spread to a standard deviation of 1

One standard deviation from the mean means the z-score is between -1 and 1.
